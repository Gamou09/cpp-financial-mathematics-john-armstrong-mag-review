```python
# load the libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import lognorm
```

```python
# Load data
prices = np.loadtxt("simulated_prices.csv")
prices
```

```python
# Parameters as C++ 
SO = 100
r = 0.05
sigma = 0.2
T = 1.0
```

```python
# Histogram of simulated prices
plt.hist(prices, bins=100, density=True, alpha=0.6, label="Simulated")
plt.show() 
```

```python
# Theoretical log normal pdf
x = np.linspace(min(prices), max(prices), 10000)
pdf = lognorm.pdf(x, s=sigma*np.sqrt(T), scale=SO*np.exp((r-0.5*sigma**2)*T))
plt.plot(x, pdf, 'r-', lw=2, label="Theoretical log-normal")
plt.show() 
```

```python
# Histogram of simulated prices
plt.hist(prices, bins=100, density=True, alpha=0.6, label="Simulated")

# Theoretical log normal pdf
x = np.linspace(min(prices), max(prices), 10000)
pdf = lognorm.pdf(x, s=sigma*np.sqrt(T), scale=SO*np.exp((r-0.5*sigma**2)*T))
plt.plot(x, pdf, 'r-', lw=2, label="Theoretical log-normal")

plt.xlabel("Price")
plt.ylabel("Density")
plt.title("Simulated Prices vs Theoretical Log-normal PDF")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
